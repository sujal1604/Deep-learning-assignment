## Task 1: Flipkart Price Sequence Dataset and LSTM Input

```python
import numpy as np

prices = np.array([999, 1049, 1099, 1079, 1149, 1199, 1249, 1229, 1299, 1349], dtype=float)

X = []
y = []

for i in range(len(prices) - 3):
    X.append(prices[i:i+3])
    y.append(prices[i+3])


X = np.array(X)
y = np.array(y)

X = X.reshape((X.shape[0], X.shape[1], 1))

print("Price sequence:", prices)
print("X shape:", X.shape)
print("y shape:", y.shape)
print("X:")
print(X)
print("y:")
print(y)
```

### Task 2: LSTM Model for Flipkart Price Prediction

```python
import numpy as np 
import tensorflow as tf 
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense

prices = np.array([999, 1049, 1099, 1079, 1149, 1199, 1249, 1229, 1299, 1349], dtype=float)

X = []
y = []

for i in range(len(prices) - 3):
    X.append(prices[i:i+3])
    y.append(prices[i+3])


X = np.array(X)
y = np.array(y)

X = X.reshape((X.shape[0], X.shape[1], 1))

model = Sequential([LSTM(32, input_shape=(3,1)), Dense(1)]) 

model.compile(optimizer="adam", loss="mse")
history = model.fit(X, y, epochs=10, verbose=1) 
```

## Task 3: Predict Next Flipkart Price

```python
last_sequence = np.array([1199, 1249, 1229], dtype=float)
last_sequence = last_sequence.reshape((1, 3, 1))

predicted_price = model.predict(last_sequence)

actual_price = 1299

print("Pridicted Price:", predicted_price[0][0])
print("Actual Price:", actual_price)
```

## Task 4: GRU Model and Training Time Comparison

```python
import time 
import numpy as np 
import tensorflow as tf 
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import GRU, Dense

gru_model = Sequential([GRU(32, input_shape=(3,1)), Dense(1)])
gru_model.compile(optimizer="adam", loss="mse")

start_time = time.time()
gru_history = gru_model.fit(X, y, epochs=10, verbose=1)

lstm_training_time = time.time() - start_time
gru_training_time = time.time() - start_time

gru_prediction = gru_model.predict(last_sequence)

print("LSTM Training Time:", lstm_training_time, "seconds")
print("GRU Training Time:", gru_training_time, "seconds")
print("LSTM Prediction:", predicted_price[0][0])
print("GRU Prediction:", gru_prediction[0][0])
print("Actual Price:", actual_price)
```

## Task 5: General Sequence Preparation Function for LSTM

```python
import numpy as np 

def prepare_lstm_data(sequence, time_steps=3):
    sequence = np.array(sequence, dtype=float)

    X = []
    y = []

    for i in range(len(sequence) - time_steps):
        X.append(sequence[i:i + time_steps])
        y.append(sequence[i + time_steps])

    X = np.array(X)
    y = np.array(y)

    X = X.reshape((X.shape[0], X.shape[1], 1))

    return X, y

spotify_plays = [120, 145, 160, 180, 175, 200, 220, 240]

X, y = prepare_lstm_data(spotify_plays, time_steps=3)

print("X shape:", X.shape)
print("y shape:", y.shape)
print("X:")
print(X)
print("y:")
print(y)
```
