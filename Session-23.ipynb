## Task 1: Load and Analyze Pretrained ResNet50

```python
import tensorflow as tf 
from tensorflow.keras.applications import ResNet50


model = ResNet50(weights="imagenet", include_top=True)

model.summary()

trainable_layers = sum(layer.trainable for layer in model.layers)

print("Total number of layers:", len(model.layers))
print("Number of trainable layers:", trainable_layers)
```

## Task 2: VGG16 Feature Extraction Using CIFAR-10

```python
import tensorflow as tf 
import numpy as np 
from tensorflow.keras.applications import VGG16
from tensorflow.keras.applications.vgg16 import preprocess_input

(x_train, y_train), (x_test, y_test) = tf.keras.datasets.cifar10.load_data()

images = x_train[:10]
images = tf.image.resize(images, (224,224))

image = images.numpy()
images = preprocess_input(images)

vgg_model = VGG16(weights = "imagenet",
                  include_top = "False",
                  input_shape = (224, 224, 3))

vgg_model.trainable = False
features = vgg_model.predict(images)

print("Original Image Shape:", images.shape)
print("Extracted features shape", features.shape)
```

## Task 3: InceptionV3 Transfer Learning

```python
import tensorflow as tf 
from tensorflow.keras.applications import InceptionV3
from tensorflow.keras.models import Model
from tensorflow.keras.layers import GlobalAveragePooling2D, Dense 

base_model = InceptionV3(weights="imagenet",
                        include_top=False,
                        input_shape=(299, 299, 3))

base_model.trainable = False
for layer in base_model.layers[-30:]:
    layer.trainable = True

x = base_model.output
x = GlobalAveragePooling2D()(x)
x = Dense(5, activation="softmax")(x)

model = Model(inputs=base_model.input, outputs=x)

trainable_params = sum(
    tf.keras.backend.count_params(weight)
    for weight in model.trainable_weights
)

print("Trainable parameters:", trainable_params)

model.summary()
```

## Task 5: InceptionV3 Transfer Lerarning With Dropout 

```python
import tensorflow as tf
from tensorflow.keras.applications import InceptionV3
from tensorflow.keras.models import Model
from tensorflow.keras.layers import GlobalAveragePooling2D
from tensorflow.keras.layers import Dense, Dropout

base_model = InceptionV3(
    weights="imagenet",
    include_top=False,
    input_shape=(299, 299, 3)
)


base_model.trainable = False
x = base_model.output
x = GlobalAveragePooling2D()(x)
x = Dense(128, activation="relu")(x)
x = Dropout(0.5)(x)


output = Dense(1, activation="sigmoid")(x)
model = Model(inputs=base_model.input, outputs=output)
model.compile(optimizer="adam", loss="binary_crossentropy", metrics=["accuracy"])


model.summary()
```
