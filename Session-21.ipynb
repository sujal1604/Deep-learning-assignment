## Task 3: Insert BatchNormalization Layer  

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, BatchNormalization, MaxPooling2D
from tensorflow.keras.layers import Flatten, Dense, Dropout

model = Sequential([
    Conv2D(32, (3,3), activation="relu", input_shape=(64, 64, 3)),

    BatchNormalization(),
    MaxPooling2D(pool_size=(2,2)),
    Flatten(),
    Dropout(0.3),

    Dense(64, activation="relu")
])

model.summary()
```

## Task 4: Compile and Train the CNN Model with Random CIFAR-10 Data

```python
import numpy as np 
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import  Conv2D, BatchNormalization, MaxPooling2D
from tensorflow.keras.layers import Flatten, Dense, Dropout 

model = Sequential([
    Conv2D(32, (3, 3), activation='relu', input_shape=(32, 32, 3)),
    BatchNormalization(),
    MaxPooling2D(pool_size=(2,2)),
    Flatten(),
    Dropout(0.3),
    Dense(64, activation="relu"),
    Dense(10, activation="softmax")
])


model.compile(optimizer="adam", loss="categorical_crossentropy", metrics=["accuracy"])

X_train = np.random.random((1000, 32, 32, 3))

y_labels = np.random.randint(0, 10, 1000) 

y_train = np.eye(10)[y_labels]

model.fit(X_train, y_train, epochs=5, batch_size=32)
```

## Task 5:   Improve CNN Architecture with Additional Convolutional Layer

```python
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, BatchNormalization
from tensorflow.keras.layers import MaxPooling2D, Flatten, Dense, Dropout

model = Sequential([
    Conv2D(32, (3, 3), activation='relu', input_shape=(32, 32, 3)),
    BatchNormalization(),
    MaxPooling2D(pool_size=(2, 2)),


    Conv2D(64, (3,3), activation="relu"),
    BatchNormalization(),
    MaxPooling2D(pool_size=(2,2)),

    Flatten(),
    Dropout(0.3),
    Dense(64, activation="relu"),
    Dense(10, activation="softmax")
])

model.compile(optimizer="adam", loss="categorical_crossentropy", metrics=["accuracy"])

X_train = np.random.random((1000, 32, 32, 3))

y_labels = np.random.randint(0, 10, 1000)

y_train = np.eye(10)[y_labels]


%history = model.fit(X_train, y_train, epochs=5, batch_size=32)

model.summary()
```
