## Task 1: Resize Images

```python
from PIL import Image
import os

os.makedirs("resized_images", exist_ok=True)

image_files = [
    "Images/Image1.png",
    "Images/Image2.png",
    "Images/Image3.png"
]

for i, file in enumerate(image_files, start=1):
    img = Image.open(file)

    resized_img = img.resize((224,224))

    resized_img.save(f"resized_images/rresized_{i}.png")

    print(f"Saved: resized_images/resized_{i}.png")

print("All images resized successfully ")
```

## Task 2: Convert Image to Grayscale Back to RGB 

```python
from PIL import Image
import matplotlib.pyplot as plt 

img = Image.open("Images/Image1.png").convert("RGB")

gray_img = img.convert("L")

rgb_img = gray_img.convert("RGB")

plt.figure(figsize=(12,4))

plt.subplot(1, 3, 1)
plt.imshow(img)
plt.title("Original Animals Images")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(gray_img, cmap="gray")
plt.title("Grayscale Images")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(rgb_img)
plt.title("RGB Image")
plt.axis("off")  

plt.show()
```

## Task 3: Image Rotation and Horizontal Flip 

```python
import numpy as np 
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.utils import img_to_array, load_img


img = load_img("Images/Image2.png")
img = img.resize((224,224))

x = img_to_array(img)
x = np.expand_dims(x, axis=0)

datagen = ImageDataGenerator(rotation_range=30, horizontal_flip=True)
augmented_images = datagen.flow(x, batch_size=1)

plt.figure(figsize=(10,8))

for i in range (4):
    batch = next(augmented_images)
    aug_img = batch[0].astype("uint8")

    plt.subplot(2, 2, i+1)
    plt.imshow(aug_img)
    plt.title(f"Augmented Image {i+1}")
    plt.axis("off")

plt.tight_layout()
plt.show()
```

## Task 4: Brightness Increase and Decrease By 40%

```python
import numpy as np
import matplotlib.pyplot as plt 
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.utils import img_to_array, load_img

img = load_img("Images/Image3.png")
img = img.resize((224,224))

x = img_to_array(img)
x = np.expand_dims(x, axis=0)

bright_datagen = ImageDataGenerator(brightness_range=[1.4, 1.4])
bright_generator = bright_datagen.flow(x, batch_size=1)

bright_img = next(bright_generator)[0].astype("uint8")

dark_datagen = ImageDataGenerator(brightness_range=[0.6, 0.6])
dark_generator = dark_datagen.flow(x, batch_size=1)

dark_img = next(dark_generator)[0].astype("uint8")

plt.figure(figsize=(12,4))

plt.subplot(1, 3, 1)
plt.imshow(x[0].astype("uint8"))
plt.title("Original")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(bright_img)
plt.title("Brightened (+40%)")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(dark_img)
plt.title("Darkened (-40%)")
plt.axis("off")

plt.tight_layout()
plt.show
```

## Task 5: Apply Shear Transformation 

```python
import numpy as np 
import matplotlib.pyplot as plt 
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.utils import img_to_array, load_img

img = load_img("Images/Image3.png")
img = img.resize((224,224))

x = img_to_array(img)
x = np.expand_dims(x, axis=0)

datagen = ImageDataGenerator(shear_range=30)
generator = datagen.flow(x, batch_size=1)

sheared_img = next(generator)[0].astype("uint8") 

plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.imshow(x[0].astype("uint8"))
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(sheared_img)
plt.title("Sheared Image")
plt.axis("off")

plt.tight_layout()
plt.show()
```
