## Task 2: Simple RNN Cell From Scratch

```python
import numpy as np 

sequence = [10, 20, 15, 30, 25]

W = 0.1
U = 0.5
b = 0.0

h = 0.0

print("Initial hidden state:", h)

for x in sequence:
    h = np.tanh(W * x + U * h + b)
    print("Hidden State:", h)

print("Final hidden state:", h)
```

## Task 3: Word-to-Index Mapping and RNN Hidden States

```python
import numpy as np 

sentence = "I love cricket and IPL matches"
words = sentence.split()

word_to_index = {word: index + 1 
                for index, word in enumerate(words)}

print("Word-to-index mapping:")
print(word_to_index)

sequence = [word_to_index[word] for word in words] 
print("\nInteger sequence:")
print(sequence)

W = 0.2
U = 0.5
b = 0.0

h = 0.0

print("\nHidden state:")

for word, x in zip(words, sequence):
    h = np.tanh(W * x + U * h + b)
    print(word, "->", x, "-> hidden state =", h)
```

LSTM and GRU networks were developed to handle long-term dependencies better and reduce this problem.

## Task 5: BPTT for a 3-Step RNN

```python
import numpy as np 

x = np.array([1.0, 2.0, 3.0])
y_true = np.array([1.0, 2.0, 3.0])

W = 0.5
U = 0.5
b = 0.0

h = 0.0

hidden_states = []

for t in range(3):
    h = np.tanh(W * x[t] + U * h + b)
    hidden_states.append(h)

y_pred = np.array(hidden_states)
loss = np.mean((y_true - y_pred) ** 2)

print("Hidden states:", hidden_states)
print("\nPridicted values:", y_pred)
print("\nLoss:", loss)

dW = 0.0
dU =0.0
db = 0.0
dh_next = 0.0

for t in reversed(range(3)):
    dy = 2 * (y_pred[t] - y_true[t]) / 3
    dh = dy + dh_next
    dtanh = dh * (1 - hidden_states[t] ** 2)

    dW += dtanh * x[t]

    if t > 0:
        previous_h = hidden_states[t - 1]
    else :
        previous_h = 0.0

    dU += dtanh * previous_h
    db += dtanh
    dh_next = dtanh * U

print("\nGradients:")
print("dW =", dW)
print("dU =", dU)
print("db =", db)
```
